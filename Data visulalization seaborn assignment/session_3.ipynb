task 1.
Import Seaborn and use sns.histplot() to visualize the distribution of delivery times (in minutes) from a sample Swiggy order dataset with at least 30 values.


```python
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd
```

```python
delivery_time = [
    20, 25, 30, 35, 28, 32, 40, 45, 22, 27,
    33, 38, 29, 31, 36, 42, 24, 26, 34, 39,
    21, 23, 37, 41, 30, 28, 35, 32, 27, 44
]

df = pd.DataFrame({
    'Delivery_Time': delivery_time
})

sns.histplot(data=df, x='Delivery_Time', bins=10)

plt.title("Swiggy Delivery Time Distribution")
plt.xlabel("Delivery Time (Minutes)")
plt.ylabel("Number of Orders")

plt.show()
```

task 2.
Download or create a list of daily step counts (e.g., from a fitness app) for one month and plot the distribution using sns.kdeplot(), showing the probability density curve.


```python
steps = [
    4500, 6000, 7500, 8000, 5500, 9000, 10000, 6500, 7000, 8500,
    5000, 7500, 9500, 11000, 6000, 8000, 7200, 8800, 9200, 6800,
    5500, 7800, 10500, 11500, 6200, 8400, 7600, 9800, 8900, 7000
]

df = pd.DataFrame({
    'Daily_Steps': steps
})

sns.kdeplot(data=df, x='Daily_Steps', fill=True)

plt.title("Daily Step Count Distribution")
plt.xlabel("Number of Steps")
plt.ylabel("Probability Density")

plt.show()
```

task 3.
Combine sns.histplot() and sns.kdeplot() to visualize the distribution of monthly spending amounts (in rupees) on Zomato orders, using the same dataset for both plots on a single figure.<br><br><em><strong>Hint:</strong> Use the same data and overlay the KDE curve on the histogram by calling both functions on the same axes.</em>


```python
spending = [
    1200, 1500, 1800, 2000, 2500, 1700, 2200, 3000, 2800, 1900,
    1600, 2100, 2400, 2600, 3200, 3500, 2300, 2700, 1800, 2900,
    3100, 2000, 1500, 2500, 3300, 2800, 2200, 1700, 2600, 3000
]

df = pd.DataFrame({
    'Monthly_Spending': spending
})

sns.histplot(
    data=df,
    x='Monthly_Spending',
    bins=10,
    stat='density',
    color='skyblue'
)

sns.kdeplot(
    data=df,
    x='Monthly_Spending',
    color='red'
)

plt.title("Zomato Monthly Spending Distribution")
plt.xlabel("Monthly Spending (₹)")
plt.ylabel("Density")

plt.show()
```
